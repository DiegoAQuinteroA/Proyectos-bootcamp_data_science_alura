```python
parametros
```

### .

### Definiendo la función

```python
def modelo_resceta(x_h, x_l):
    # limites entre la normalizacion
    limite_normalizados = [-1,+1]
    limite_harina = [0.5,1.5]
    limite_leche = [0.2,0.7]

    # convertir
    x_h_convertido = np.interp(x_h,limite_harina,limite_normalizados)
    x_l_convertido = np.interp(x_h,limite_leche,limite_normalizados)

    cantidades = parametros['Intercept'] + (parametros['Harina']*x_h_convertido) + (parametros['Leche']*x_l_convertido)
    return round(cantidades)
```

```python
modelo_resceta(1,0.5)
```

```python
x_harina = np.linspace(start=0.5, stop=1.5, num=10)
x_harina
```

```python
x_leche = np.linspace(start=0.2, stop=0.7, num=10)
x_leche
```

### .

```python
puntos = []
for cont1 in x_harina:
    temp=[]
    for cont2 in x_leche:
        temp.append(modelo_resceta(cont1, cont2))
    puntos.append(temp)
```
