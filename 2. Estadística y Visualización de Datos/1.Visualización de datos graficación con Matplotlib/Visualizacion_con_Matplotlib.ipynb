```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# agrupar por la temperatura media de los dias de la semana
temporatura_por_dia_de_semana= df.groupby('dia_de_semana')['temperatura'].mean()
nombres_dias = ['Domingo', 'Lunes', 'Martes', 'Miercoles', 'Jueves', 'Viernes', 'Sabado']
temporatura_por_dia_de_semana = temporatura_por_dia_de_semana[nombres_dias]
temporatura_por_dia_de_semana
```

```python
# generar graficos de barras
fig = plt.figure(figsize=(5,4))
eje = fig.add_axes([0,0,1,1])
# crear un indice para cada dia de la semana
indice = range(len(temporatura_por_dia_de_semana))

eje.bar(indice, temporatura_por_dia_de_semana)
eje.set_title('Temperatura por dias de semana', fontsize=15, pad=10)
eje.set_xlabel('Dia de semana', fontsize=15)
eje.set_ylabel('temperatura', fontsize=15)
```

```python
# generar graficos de barras
fig = plt.figure(figsize=(5,4))
eje = fig.add_axes([0,0,1,1])
# crear un indice para cada dia de la semana
indice = range(len(temporatura_por_dia_de_semana))

eje.bar(indice, temporatura_por_dia_de_semana)
eje.set_title('Temperatura por dias de semana', fontsize=15, pad=10)
eje.set_xlabel('Dia de semana', fontsize=15)
eje.set_ylabel('temperatura', fontsize=15)
# agragar labels/etiquetas
eje.set_xticks(indice) #asignar que quiero de label
eje.set_xticklabels(nombres_dias) #signar nombres a los indices anteriores
```

##  Agregando anotaciones y líneas horizontales

```python
fig = plt.figure(figsize=(5,4))
eje = fig.add_axes([0,0,1,1])

# media y desviacion estandar
mu, sigma = df2['PetalLengthCm'].mean(), df2['PetalLengthCm'].std()

# crear grafico + numero columnas
eje.hist(df2['PetalLengthCm'], bins=20)
eje.set_title('Histograma', fontsize=15, pad=10)
eje.set_xlabel('PetalLengthCm', fontsize=15)

eje.grid(True)

# anotacion
eje.annotate('$mu = {0:.2f}$\n$sigma = {1:.2f}$'.format(mu, sigma),
             xy=(4.5, 20), fontsize=20)
```

```python
# crear lineas verticales para la media

fig = plt.figure(figsize=(5,4))
eje = fig.add_axes([0,0,1,1])

# media y desviacion estandar
mu, sigma = df2['PetalLengthCm'].mean(), df2['PetalLengthCm'].std()

# crear grafico + numero columnas
eje.hist(df2['PetalLengthCm'], bins=20)
eje.set_title('Histograma', fontsize=15, pad=10)
eje.set_xlabel('PetalLengthCm', fontsize=15)

eje.grid(True)

# anotacion
eje.annotate('$mu = {0:.2f}$\n$sigma = {1:.2f}$'.format(mu, sigma),
             xy=(4.5, 20), fontsize=20)
eje.axvline(mu, color='k', linestyle='--')
eje.annotate('promedio', xy=(mu - 1.7, 28), fontsize=20)
```

```python
fig = plt.figure(figsize=(5,4))
eje = fig.add_axes([0,0,1,1])

# media y desviacion estandar
mu, sigma = df2['PetalLengthCm'].mean(), df2['PetalLengthCm'].std()

# crear grafico + numero columnas
eje.hist(df2['PetalLengthCm'], bins=20)
eje.set_title('Histograma', fontsize=15, pad=10)
eje.set_xlabel('PetalLengthCm', fontsize=15)

eje.grid(True)

# anotacion y mostar media y desviacion
eje.annotate('$\mu = {0:.2f}$\n$\sigma = {1:.2f}$'.format(mu, sigma),
             xy=(4.5, 20), fontsize=20)
#mostrar la media/promedio en el grafico con una linea vertical
eje.axvline(mu, color='k', linestyle='--')
eje.annotate('promedio', xy=(mu - 1.7, 28), fontsize=20)

#mostrar la mediana
eje.axvline(df2['PetalLengthCm'].median(), color='g', linestyle='--')
eje.annotate('mediana', xy=(df2['PetalLengthCm'].median(), 31), fontsize=20)
```
